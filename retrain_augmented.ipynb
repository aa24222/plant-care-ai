````python
# Cell 1: Setup
from ultralytics import YOLO
import torch

# Check GPU
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Using device: {device}")

if device == 'cuda':
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print("Training will be fast!")
else:
    print("Warning: Training on CPU will be slow (2-3 hours)")
    print("Consider using Google Colab with free GPU")
````

````python
import torch

print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"CUDA version: {torch.version.cuda}")

if torch.cuda.is_available():
    print(f"GPU detected: {torch.cuda.get_device_name(0)}")
    print("\nReady to train on GPU!")
else:
    print("\nGPU not detected - may need to restart Jupyter completely")
````

````python
# Cell 2: Load model
model = YOLO('yolov8n-cls.pt')
````

````python
# Cell 3: Train with augmentation
results = model.train(
    data='datasets/plantvillage_yolo',
    epochs=50,
    imgsz=224,
    batch=16,
    patience=10,
    
    # Augmentation
    hsv_h=0.02,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=30,
    translate=0.2,
    scale=0.5,
    shear=10,
    perspective=0.0005,
    flipud=0.5,
    fliplr=0.5,
    mosaic=0.8,
    mixup=0.2,
    copy_paste=0.3,
    
    device='cuda',
    project='runs/classify',
    name='plant_disease_v2_augmented',
    plots=True,
    verbose=True
)
```

---

## What to Expect

With your RTX 4060, training should be **fast**:

- **Time per epoch:** ~2-3 minutes
- **Total time:** ~1-1.5 hours (for 50 epochs, but will likely stop early with patience)
- **Speed:** 10-20x faster than CPU!

You'll see output like:
```
Epoch 1/50: 100%|██████████| 1238/1238 [02:15<00:00, 9.14it/s]
               train/loss: 1.234, val/accuracy: 45.2%
               
Epoch 2/50: 100%|██████████| 1238/1238 [02:12<00:00, 9.35it/s]
               train/loss: 0.876, val/accuracy: 67.8%
...
````
